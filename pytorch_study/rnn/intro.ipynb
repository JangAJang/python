# RNN(Recursive Neural Network)

- 시계열 데이터를 위해 만들어진 Neural Network이다. 
- 단어를 주더라도, 단어의 순서가 의미가 있다. (같은 단어여도 문장에서 어디에 배치하냐에 따라 벡터가 달라질 수 있다.)
- 그래서 데이터의 벡터만 저장하는 것이 아니라, 앞에 position index를 추가한다. 
    - 단순히 추가하는 것이 아니라, 어떻게 담아야 잘 담을지 정의해야 한다.

![rnn_structure](img/rnn_structure.png)

- x0에서 h0를 반환하고, hidden state를 다음 셀로 전달한다.
- 결국 n번째 데이터는 n-1번째 데이터에 영향을 받게 된다.
- `HELLO`라는 것을 입력했을 때, 두 개의 `L`에서 hidden state를 통해 다음이 `O`인지, `L`인지 구분하게 된다. 
- 셀 `A`는 항상 하나다. 

## 셀 A에서 일어나는 일

![cell_a](img/cell_a.png)

- 기본적으로는 이전 단계의 hidden state, 이번 단계의 입력값으로 출력 h를 만든다. 
- 함수가 어떻게 될지는 설계에 따라 달라진다. 
- 이중에 유명한게 LSTM, GRU등이 존재한다. 
- A가 복잡할 수록, 학습 수준이 높아지지만 자원이 많이 필요해진다.(trainability가 증가한다)
