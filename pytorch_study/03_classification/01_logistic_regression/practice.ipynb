## 문제 정의
- 타이타닉 승객의 특징(객실 등급 `Pclass`, 성별 `Sex`, 나이 `Age`, 운임 `Fare`, 동반 가족 수 등)을 입력으로 받아 **생존 여부(`Survived` = 0/1)** 를 예측한다.
- 입력: 전처리된 수치/범주 특징 벡터. 출력: 생존 확률 $P(\text{survived}=1\mid x)$, 임계값 0.5로 0/1 라벨 결정.
