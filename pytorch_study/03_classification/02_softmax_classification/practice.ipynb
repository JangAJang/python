## 문제 정의
- 28x28 픽셀(784차원) 손글씨 숫자 이미지를 입력으로 받아 **0~9 중 어떤 숫자인지(10개 클래스)** 예측한다.
- 입력: 784차원 픽셀 벡터(0~255, 보통 255로 정규화). 출력: 10개 클래스에 대한 로짓 → 소프트맥스 확률 → argmax로 클래스 결정.
- 목표: 소프트맥스 + 교차 엔트로피로 다중 분류기를 학습해 검증 정확도를 높인다.
