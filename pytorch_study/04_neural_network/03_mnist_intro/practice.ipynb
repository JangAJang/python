## 문제 정의
- 입력: 28x28 grayscale 손글씨 숫자 이미지를 784차원 벡터로 펼친 것.
- 목표: `Linear(784, 10) + CrossEntropyLoss(softmax 내장)`로 0~9 (10개 클래스)를 분류한다.
- 출력: 테스트셋 정확도(단층 선형 모델 기준 약 88~90%)와 샘플 1장의 예측 결과.
